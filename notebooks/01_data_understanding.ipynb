# Unique values of each categorical column (spot inconsistent labels)

```python

for col in df.select_dtypes("object").columns.drop(["customerID", "TotalCharges"]):
    print(col, "->", df[col].unique())
```

# TotalCharges is stored as text, so check for hidden bad values

```python

tc = pd.to_numeric(df["TotalCharges"], errors="coerce")
print("Non-numeric TotalCharges:", tc.isna().sum())
df.loc[tc.isna(), ["customerID", "tenure", "MonthlyCharges", "TotalCharges"]]
```
